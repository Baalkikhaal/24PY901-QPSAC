# Time independent Schrodinger equation

Let us derive the time independent Schrodinger equation for two cases -

- finite dimensional Hilbert basis
- infinite dimensional Hilbert basis

# Finite dimensional Hilbert basis

The aim is to solve the Time Dependent Schrodinger equation (TDSE)

$$
\imath \hbar \frac{\mathrm{d} |\psi\rangle}{\mathrm{d} t} = \tilde{H} |\psi\rangle
$$

## Stationary states

A stationary state $|\psi\rangle$ is an eigenstate of the Hamiltonian $\tilde{H}$ so that

$$
\tilde{H} |\psi\rangle = E |\psi\rangle,
$$

where $E$ is the energy eigenvalue. This eigenvalue equation is called Time Independent Schrodinger Equation (TISE).



Let us consider the computational basis as the energy eigenbasis $\{|E_i\rangle\}$ where $i = \{1, 2, \ldots, n\}$, where $n$ is the dimension of the Hilbert space. Let us consider the initial state $(t = 0)$ as a linear combination of the energy eigenkets as

$$
|\psi(0)\rangle
=
\left(
\begin{matrix}
c_1(0) \\
c_2(0) \\
\vdots \\
c_n(0) \\
\end{matrix}
\right)
$$

Since the Hamiltonian is diagonal in its own eigen basis, it is given by

$$
\tilde{H}
=
\left(
\begin{matrix}
E_1 & 0 & \ldots & 0 \\
0 & E_2 & \ldots & 0 \\
\vdots & \vdots & \ddots & \vdots \\
0 & 0 & \ldots & E_n \\
\end{matrix}
\right)
$$

Let us consider the general state at $t$ as

$$
|\psi(t)\rangle
=
\left(
\begin{matrix}
c_1(t) \\
c_2(t) \\
\vdots \\
c_n(t) \\
\end{matrix}
\right)
$$

Upon substitution of $|\psi(t)\rangle$ and $\tilde{H}$ in TDSE, we have

$$
\imath \hbar
\left(
\begin{matrix}
\frac{\mathrm{d} c_1(t)}{\mathrm{d} t} \\
\frac{\mathrm{d} c_2(t)}{\mathrm{d} t} \\
\vdots \\
\frac{\mathrm{d} c_n(t)}{\mathrm{d} t} \\
\end{matrix}
\right)
=
\left(
\begin{matrix}
E_1 c_1(t) \\
E_2 c_2(t) \\
\vdots \\
E_n c_n(t) \\
\end{matrix}
\right)
$$

We have a set of $n$ first order differential equations

\begin{align}
\imath \hbar\frac{\mathrm{d} c_1(t)}{\mathrm{d} t} &= E_1 c_1(t) \\
\imath \hbar\frac{\mathrm{d} c_2(t)}{\mathrm{d} t} &= E_2 c_2(t) \\
\vdots &= \vdots \\
\imath \hbar\frac{\mathrm{d} c_n(t)}{\mathrm{d} t} &= E_n c_n(t) \\
\end{align}

This set is reduced to

\begin{align}
\frac{\mathrm{d} c_1(t)}{\mathrm{d} t} &= -\frac{\imath E_1}{\hbar} c_1(t) \\
\frac{\mathrm{d} c_2(t)}{\mathrm{d} t} &= -\frac{\imath E_2}{\hbar} c_2(t) \\
\vdots &= \vdots \\
\frac{\mathrm{d} c_n(t)}{\mathrm{d} t} &= -\frac{\imath E_n}{\hbar} c_n(t) \\
\end{align}

The solutions to these differential equations have exponential form with the initial values serving as pre-factors

\begin{align}
c_1(t) &=  c_1(0) \exp \left(-\frac{\imath E_1 t}{\hbar}\right) \\
c_2(t) &=  c_2(0) \exp \left(-\frac{\imath E_2 t}{\hbar}\right) \\
\vdots &= \vdots \\
c_n(t) &=  c_n(0) \exp \left(-\frac{\imath E_n t}{\hbar}\right) \\
\end{align}



The left hand side can be grouped together to represent the state at $t$ whereas the right hand side is the solution of the TDSE given by

$$
|\psi(t)\rangle
=
\left(
\begin{matrix}
c_1(t) \\
c_2(t) \\
\vdots \\
c_n(t) \\
\end{matrix}
\right)
=
\left(
\begin{matrix}
c_1(0) \exp \left(-\frac{\imath E_1 t}{\hbar}\right) \\
c_2(0) \exp \left(-\frac{\imath E_2 t}{\hbar}\right) \\
\vdots \\
c_n(0) \exp \left(-\frac{\imath E_n t}{\hbar}\right) \\
\end{matrix}
\right)
$$


This can be decomposed into a linear combination of energy eigenkets

\begin{align}
|\psi(t)\rangle &= c_1(0) \exp \left(-\frac{\imath E_1 t}{\hbar}\right) |E_1\rangle + c_2(0) \exp \left(-\frac{\imath E_2 t}{\hbar}\right) + \ldots + c_n(0) \exp \left(-\frac{\imath E_n t}{\hbar}\right), \\
&= \sum_{i} c_i(0) \exp \left(-\frac{\imath E_i t}{\hbar}\right) |E_i\rangle.
\end{align}

$$
Q.E.D.
$$

# TISE in the coordinate basis

We have seen that the solution of the TDSE is intimately connected to the TISE

$$
\tilde{H} |\psi\rangle = E |\psi\rangle,
$$

where the Hamiltonian can be written in terms of the position $\tilde{X}$ and momentum $\tilde{P}$ operators as

$$
\tilde{H} (\tilde{X}, \tilde{P}) = \frac{\tilde{P}^2}{2m} + V(\tilde{X}).
$$

Here, $\tilde{T} = \tfrac{\tilde{P}^2}{2m}$ is the kinetic energy operator and $\tilde{V}(\tilde{X})$ is the potential energy operator.

Thus, we can express the Hamiltonian operator in the coordinate if we can express the position and momentum operators in the coordinate basis. The matrix elements of the operators are given by

\begin{align}
\langle x | \tilde{X} | x' \rangle &= x \delta(x-x'), \\
\langle x | \tilde{P} | x' \rangle &= \frac{\hbar}{\imath} \delta(x-x')\frac{\mathrm{d}}{\mathrm{d}x}.
\end{align}

As a result, the matrix elements of operators $\tilde{T}$ and $\tilde{V}$ are given by

\begin{align}
\langle x | \frac{\tilde{P}^2}{2m} | x' \rangle &= -\frac{\hbar^2}{2m} \delta(x-x')\frac{\mathrm{d^2}}{\mathrm{d}x^2} \\
\langle x | \tilde{V}(\tilde{X}) | x' \rangle &= V(x) \delta(x-x')
\end{align}

As we are working in the coordinate basis, the completeness relation is given by

$$
\tilde{I} = \int_{-\infty}^{\infty} |x'\rangle \langle x' | \, \mathrm{d} x'
$$

Let us plug the completeness relation inside the TISE, so that

\begin{align}
\tilde{H} |\psi\rangle &= E |\psi\rangle, \\
\Leftrightarrow \tilde{H} \tilde{I}|\psi\rangle &= E |\psi\rangle, \\
\Leftrightarrow  \tilde{H} \left( \int_{-\infty}^{\infty} |x'\rangle \langle x' | \, \mathrm{d} x' \right) |\psi\rangle &= E |\psi\rangle, \\
\Leftrightarrow \int_{-\infty}^{\infty} \tilde{H}  |x'\rangle \langle x' | |\psi\rangle \, \mathrm{d} x'   &= E |\psi\rangle, \\
\end{align}

Take the inner product with $\langle x | $ on both sides of the equation so that

\begin{align}
 \left( \int_{-\infty}^{\infty} \tilde{H}  |x'\rangle \langle x' | |\psi\rangle \, \mathrm{d} x' \right)  &= E \langle x |\psi\rangle, \\
\int_{-\infty}^{\infty} \langle x | \tilde{H}  |x'\rangle \langle x' | |\psi\rangle \, \mathrm{d} x' & = E \psi (x). \tag{A}
\end{align}

Here, we have used the definition of wavefunction $\psi (x) := \langle x |\psi\rangle$.

Now,

\begin{align}
\langle x | \tilde{H}  |x'\rangle &= \langle x | \tilde{T}  + \tilde{V} |x'\rangle, \\
&= -\frac{\hbar^2}{2m} \delta(x-x')\frac{\mathrm{d^2}}{\mathrm{d}x^2} + V(x) \delta(x-x'), \\
&= \left( -\frac{\hbar^2}{2m}\frac{\mathrm{d^2}}{\mathrm{d}x^2} + V(x) \right) \delta(x-x')
\end{align}

Therefore, this is the matrix element of the Hamiltonian operator in the coordinate basis.

If we use the matrix element in the integral in Equation. (A), then

$$
\int_{-\infty}^{\infty} \left( -\frac{\hbar^2}{2m}\frac{\mathrm{d^2}}{\mathrm{d}x^2} + V(x) \right) \delta(x-x') \psi (x') \, \mathrm{d} x' = E \psi (x).
$$

Here, we have again used the definition of wavefunction $\psi (x') := \langle x' |\psi\rangle$.

Using the definition of delta function

$$
\int_{-\infty}^{\infty} f(x')\delta(x-x')\, \mathrm{d} x' = f(x),
$$

we have

$$
\left( -\frac{\hbar^2}{2m}\frac{\mathrm{d^2}}{\mathrm{d}x^2} + V(x) \right)\psi (x) = E \psi (x)
$$

This is the TISE in the coordinate basis

$$
Q.E.D.
$$
