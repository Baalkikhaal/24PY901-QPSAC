# Wave ket postulate

A quantum system is completely described by a normalized state vector |ψ⟩.

# Position representation

ψ(x) = ⟨x|ψ⟩

# Normalization

∫ |ψ(x)|² dx = 1

```python
import numpy as np

x = np.linspace(-5,5,1000)
psi = np.exp(-x**2)
norm = np.trapz(np.abs(psi)**2, x)
norm
```
