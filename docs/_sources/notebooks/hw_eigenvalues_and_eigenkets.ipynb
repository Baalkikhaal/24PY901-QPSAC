# Eigen values and eigen kets of $\tilde{S}_x$

The action of $\tilde{S}_x$ is given by

$$
			\begin{align*}
				\tilde{S}_x \ket{0} &= \frac{\hbar}{2} \ket{1}	\\
				\tilde{S}_z \ket{1} &= \frac{\hbar}{2} \ket{0}	\\				
			\end{align*}
$$

Find the eigenvalues and eigenkets of $\tilde{S}_x$.

# Solution

The matrix representation of the computational basis $\{ \ket{0}, \ket{1}\}$ is given by

$$
\ket{0} = 
\left(
\begin{matrix}
    1 		\\
    0 	
\end{matrix}
\right)
,
\qquad
\ket{1} = 
\left(
\begin{matrix}
    0 		\\
    1 	
\end{matrix}
\right)
$$

In this basis, the action on the basis is given by

$$
\tilde{S}_x\ket{0} 
=
\frac{\hbar}{2}
\left(
\begin{matrix}
    0 		\\
    1 	
\end{matrix}
\right)
,
\qquad
\tilde{S}_x\ket{1} 
=
\frac{\hbar}{2}
\left(
\begin{matrix}
    1 		\\
    0 	
\end{matrix}
\right)
$$

Therefore, the matrix representation of $\tilde{S}_x$ is given by

$$
			\tilde{S}_x
			=
			\frac{\hbar}{2}
			\left(
			\begin{matrix}
				0 	&	1	\\
				1 	&	0
			\end{matrix}
			\right)
$$

# Eigen values

The eigen values are given by solutions to the determinant equation

$$
\left|
\begin{matrix}
    -\lambda 	&	1	\\
    1 	&	-\lambda
\end{matrix}
\right|
=
0.
$$

$$
\lambda^2 = 1 \Leftrightarrow \lambda = \pm 1
$$

# Eigen kets

Case (I): $\lambda = + 1$

$$
\left(
\begin{matrix}
    -1 	&	1	\\
    1 	&	-1
\end{matrix}
\right)
\left(
\begin{matrix}
    \alpha	\\
    \beta
\end{matrix}
\right)
=
0
$$

$$
\alpha = \beta
$$

So the eigen ket is parallel to

$$
\ket{+1} = \left(
\begin{matrix}
    1	\\
    1
\end{matrix}
\right)
$$


Since the eigen ket should have norm equal to one, 

$$
\ket{+1} 
=
\frac{1}{\sqrt{2}}
\left(
\begin{matrix}
    1	\\
    1
\end{matrix}
\right)
$$

Case (II): $\lambda = - 1$

$$
\left(
\begin{matrix}
    1 	&	1	\\
    1 	&	1
\end{matrix}
\right)
\left(
\begin{matrix}
    \alpha	\\
    \beta
\end{matrix}
\right)
=
0
$$

$$
\alpha = -\beta
$$

So the normalized eigen ket is

$$
\ket{-1} 
=
\frac{1}{\sqrt{2}}
\left(
\begin{matrix}
    1	\\
    -1
\end{matrix}
\right)
$$
