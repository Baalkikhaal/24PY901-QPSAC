# Coherence

Now that we understand how the light is absorbed or emitted in quantum of radiation, let us look at the light from a classical physics perspective

The model of light under classical perspective is that light is an electromagnetic wave with transverse electric and magntetic fiels. the direction of propagation is perpendicular to both the fields. 

This property is captured mathematically using vector algebra as

$$
\vec{k} || \vec{E} \times \vec{B}
$$

# Coherence

So the electric and magnetic fields have two features

- wave nature of the fields
- transverse nature of the fields

Due the wave nature, the properties of light are governed the wave properties

- wavelength
- frequency
- amplitude
- phase
- principle of superposition

The simples wave that can be imagined is a monochromatic plane wave. This assumes the electron has emitted from the excited to the ground state and we are studying the light at a distance very far from the electron. This can be mathematically represented as 

$$
\psi (y, t) = A \exp(\overline{k y - \omega t})
$$



# Incoherent light

However, if we consider bunch of electrons emitting the light, then not all the waves may be emitted in phase. 

Since the waves follow the principle of superposition, the net amplitude of light is given by the sum of the individual amplitudes weighted by their **relative** phases. This results in incoherent light

Incoherent light is defined as the light with no definite phase relationship between any two adjacent points in space and time.

# Coherent light

If the electrons emit light in phase, then the individual waves have a phase relationship. The net amplitude is non zero, and the resultant wave will have definite phase relationship between two adjacent points in space and time. This light is called a coherent light.

# Spatial coherence 

Both of the two cases are limiting cases. In practicality, the situation is somewhere in between. To measure the degree of coherence for such lights, we introduce the concepts of

- spatial coherence
- temporal coherence

# Spatial coherence

The spatial coherence is a measure of how two points on wave that are spatially separated have a definite phase relationship.

What we mean by definite phase relationship is that the phase difference between the points should not change in other configuration axis, the time axis i.e.

If two points A and B spatially separated on the wave have phases $\phi_A$ and $\phi_B$ respectively, then the phase differnce
between them is

$$
\Delta \phi = \phi_A - \phi_B
$$

Then, we say the points A and B are spatially coherent when

$$
\Delta \phi(t) = \text{constant}
$$

# Coherence length

It is understandable that closely spaced points are likely to be spatially coherent than far apart points.

Coherence length $l_c$ is the maximum separation between points upto which the phase difference does not change by $\pi$ in a time scale defined by the velocity of the wave

If the velocity is $v$, then the time scale is

$$
\tau = \frac{v}{l_c}
$$
