# Visualizing Quantum States

So far, we have:
- Built circuits
- Computed statevectors
- Understood operators

But quantum states are abstract.

👉 Visualization helps us build intuition.

In this notebook, we will:
- Visualize quantum states on the Bloch sphere
- Plot measurement probabilities
- Connect visuals to physical meaning

## Required Tools

We use Qiskit visualization functions.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector, plot_histogram
```

## The Bloch Sphere

A single qubit state can be represented as a point on a sphere.

Key states:
- |0⟩ → north pole
- |1⟩ → south pole
- superpositions → points in between

```python
# |0> state
sv0 = Statevector.from_label('0')

plot_bloch_multivector(sv0)
```

## Observation

The vector points along the +Z direction.

👉 This represents |0⟩

```python
# |1> state
sv1 = Statevector.from_label('1')

plot_bloch_multivector(sv1)
```

## Observation

The vector points along the -Z direction.

👉 This represents |1⟩

## Superposition (H Gate)

```python
qc = QuantumCircuit(1)
qc.h(0)

sv_h = Statevector.from_label('0').evolve(qc)

plot_bloch_multivector(sv_h)
```

## Interpretation

The vector lies on the equator.

👉 Equal probability of |0⟩ and |1⟩

This is a balanced superposition.

## Phase Matters

Different states can have same probabilities but different phases.

Let us apply Z gate after H.

```python
qc2 = QuantumCircuit(1)
qc2.h(0)
qc2.z(0)

sv_hz = Statevector.from_label('0').evolve(qc2)

plot_bloch_multivector(sv_hz)
```

## Key Insight

- Probabilities remain the same
- Direction changes on Bloch sphere

👉 Phase is invisible in measurement but visible in state geometry

## Measurement Visualization (Histogram)

We now connect state → probabilities → outcomes

```python
# Probabilities from H state
probs = sv_h.probabilities_dict()

plot_histogram(probs)
```

## Interpretation

- Bars represent probabilities
- Heights correspond to likelihood of outcomes

👉 This predicts measurement results
