# Operator: Matrix Representation of Quantum Circuits

So far, we have learned:

- QuantumCircuit → describes operations
- Statevector → shows resulting state
- Measurement → gives outcomes

Now we ask:

👉 What is the mathematical form of a quantum circuit?

Answer:
👉 A quantum circuit is a matrix (Operator)

In this notebook, we will:
- Represent circuits as matrices
- Apply matrices to quantum states
- Verify consistency with Statevector

## From Gates to Matrices

Every quantum gate corresponds to a matrix.

Examples:
- X gate → flips states
- H gate → creates superposition

A circuit is a combination (multiplication) of these matrices.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, Operator
```

## Creating a Simple Circuit

```python
qc = QuantumCircuit(1)
qc.x(0)

qc
```

```python
qc.draw('mpl')
```

## Converting Circuit to Operator

```python
op = Operator(qc)

op
```

```python
list_1d = [1, 3]

list_1d
```

```python
# nesting 
list_2d = [[2, 3], [4, 5]]

list_2d
```

## Interpretation

The operator is a matrix:

[[0, 1],
 [1, 0]]

Let us write the X operator neatly using LaTeX notation

$$
\tilde{X} =
\left(
\begin{matrix}
0 & 1 \\
1 & 0 \\
\end{matrix}
\right)
$$
👉 This is exactly the X gate matrix.

So:
QuantumCircuit → Operator (matrix)

## Applying Operator to a State

```python
# Initial state |0>
sv = Statevector.from_label('0')

# Apply operator
sv_new = sv.evolve(op)

sv_new
```

```python
sv_new_2 = sv.evolve(qc)

sv_new_2
```

## Key Idea

Applying:
- circuit to statevector
- operator to statevector

should give the SAME result.

```python
# Using circuit
sv1 = sv.evolve(qc)

# Using operator
sv2 = sv.evolve(op)

sv1, sv2
```

## Observation

Both results are identical.

👉 Circuit = Operator (mathematically)

## Combining Gates

Let us apply two gates:
H followed by X

```python
qc2 = QuantumCircuit(1)
qc2.h(0)
qc2.x(0)

op2 = Operator(qc2)

op2
```

```python
# matrix multiplication
import numpy as np
matrix_h = np.array([[1/np.sqrt(2), 1/np.sqrt(2)], [1/np.sqrt(2), -1.0*1/np.sqrt(2)]])
matrix_x = np.array([[0.0, 1.0],[1.0, 0.0]])

matrix_qc = matrix_x @ matrix_h
matrix_qc

np.allclose(op2.to_matrix(), matrix_qc)
```

```python
op2.to_matrix()
```

## Important Insight

Matrix multiplication follows order:

👉 Last gate applied appears FIRST in matrix multiplication

This is subtle but important.
