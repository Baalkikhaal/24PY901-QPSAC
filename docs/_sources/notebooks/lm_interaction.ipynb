# Light matter interaction

The problem of interaction of light and matter defines more or less the problem of physics because physics deals with the working of Nature; and Nature can be classified into light and matter.

After the advent of quantum mechanics, and the concept of quantization of energy levels, the interaction of light with matter can be broadly categorized into three processes -

- Stimulated absorption
- Spontaneous emission
- Stimulated emission

Stimulated absorption forms the basis of absorption spectra, whereas spontaneous absorption forms the basis of emission spectra.
The stimulated emission forms the basis of lasers and its discussion is out of the scope of this course.
