# Polarization

Due to the transverse nature of light, the light properties are governed by 

- polarization

The polarization of light is a vector that is parallel to the electric field.

The polarization plane is the plane containing the electric field and is perpendicular to the direction of propagation.

# States of polarization

Depending on the relation of polarization with time, the classical light is classified into

- linearly polarized light
- circularly polarized light
- elliptically polarized light
