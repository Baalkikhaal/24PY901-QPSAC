# Waveket representation of polarization states

Write the wave ket representation of polarization states

- circularly,
- elliptically polarized light


# Solution

The circularly polarized light can be modelled as two components - the horizontal and vertical components. However, the vertical component is ahead in phase by $\pi/2$ compared to horizontal component.

Also, the amplitude weightage is same.
Therefore, 

$$
\begin{align}
|C_+\rangle &= \frac{1}{\sqrt{2}}|H\rangle + \exp(\imath \frac{\pi}{2})\frac{1}{\sqrt{2}}|V\rangle \\
&= \frac{1}{\sqrt{2}}(|H\rangle + \imath|V\rangle) 
\end{align}
$$

This is a right circularly polarized light. Similarly, we have left circularly polarized light...

$$
\begin{align}
|C_-\rangle &= \frac{1}{\sqrt{2}}|H\rangle + \exp(-\imath \frac{\pi}{2})\frac{1}{\sqrt{2}}|V\rangle \\
&= \frac{1}{\sqrt{2}}(|H\rangle - \imath|V\rangle) 
\end{align}
$$

In the matrix representation, 

$$
|C_\pm\rangle
\left(
\begin{array}{l}
1 \\
\pm\imath
\end{array}
\right)
$$


For the elliptically polarized light, the amplitudes are different.
