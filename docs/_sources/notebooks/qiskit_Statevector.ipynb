# Statevector: Representing Quantum States

In the previous notebook, we built quantum circuits.

But an important question remains:

👉 What is the *state* of the qubit after applying gates?

In this notebook, we will:
- Represent quantum states using vectors
- Apply circuits to states
- Observe how gates change the state

This is where physics and linear algebra enter.

## From Circuits to States

A quantum circuit describes operations.

To *see the result*, we use:

👉 Statevector

This object stores the amplitudes of a quantum system.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

## Initial State

By default, a qubit starts in:

|0⟩

We can create this using Statevector.

```python
# Create |0> state
sv = Statevector.from_label('0')

sv
```

## Interpretation

The statevector contains amplitudes:

[1, 0]

Meaning:
- Probability of |0⟩ = 1
- Probability of |1⟩ = 0

## Applying a Circuit to a State

We now:
1. Create a circuit
2. Apply it to the statevector

```python
qc = QuantumCircuit(1)
qc.x(0)
```

```python
qc.draw('text')
```

```python
# Apply circuit to state
sv_new = sv.evolve(qc)

sv_new
```

## Observation

After applying X gate:

|0⟩ → |1⟩

Statevector becomes:
[0, 1]

👉 The circuit changed the state.

## Superposition using H Gate

Now let us apply the Hadamard (H) gate.

```python
qc_h = QuantumCircuit(1)
qc_h.h(0)
sv_h = sv.evolve(qc_h)

sv_h
```

## Interpretation

The result is approximately:

[1/√2, 1/√2]

Meaning:
- 50% chance of |0⟩
- 50% chance of |1⟩

👉 This is called superposition.

## Key Idea

- QuantumCircuit → describes operations
- Statevector → shows the result

The circuit does NOT store the state.
The statevector does.
