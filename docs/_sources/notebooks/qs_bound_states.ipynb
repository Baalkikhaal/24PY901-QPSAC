# 1. Infinite potential

## Analytical solution
The analytical solution for TISE of particle in an infinite potential well has pedagogical importance.

Let us consider the wavefunction to have the form

$$
\psi(x) = A \sin k x + B \cos k x
$$

At the boundary $x = 0$, the wavefunction must vanish so that

$$
\psi(0) = 0 \Leftrightarrow B = 0
$$

Similarly, at the other boundary $x=a$, the wavefunction must vanish so that

$$
\psi(a) = 0 \Leftrightarrow \sin k a = 0 \Leftrightarrow k = \frac{n \pi}{a}, \quad n = 1, 2, 3, \ldots
$$

For each $k$, there is an associated quantum number $n$, so that $k \rightarrow k_n$, and the wavefunction is given by

$$
\psi_n(x) = A \sin \frac{n \pi x}{a}
$$

Upon normalizing the wavefunction, using

$$
\int_0^a \psi_n^*(x)\psi_n(x) \, \mathrm{d}x = 1 \Leftrightarrow A = \sqrt{\frac{2}{a}}
$$

Therefore, the wavefunction is given by

$$
\psi_n(x) = \sqrt{\frac{2}{a}} \sin \frac{n \pi x}{a}
$$

### Energy eigen values

The energy of the particle is given by

$$
E = T + V
$$

Since the potential energy $V$ is zero

$$
E = T
$$

But kinetic energy is given by

$$
E_n = \frac{\hbar^2 k_n^2}{2m} = \frac{n^2 \hbar^2 \pi^2}{2m a^2} = \frac{n^2 \pi^2}{ 2a^2}
$$
