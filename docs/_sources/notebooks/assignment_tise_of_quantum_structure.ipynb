## Finite difference method

Here, we have used the **Numerov algorithm** to obtain the recurrence relation between adjacent wave function values.



## Atomic units

Also the fudamental constants are measured in atomic units. This allows to choose the step size for finite difference independent of the scale of the problem.

In the atomic units, we consider the electron mass, reduced Planck's constant and Bohr's radius each as 1 unit.

$$
m_e = 1, \quad \hbar = 1, \quad a_0 = 1
$$

# Infinite potential well
In the assignment, numerically find the eigenvalues and eigen functions of the TISE using the "wag-the-dog" or "shooting star" method

![Ground state function and energy of infinite well](./images/infinite_potential_ground_state_wave_function_and_eigen_energy.png)

The infinite potential function is given by

$$
V(x)
=
\left\{
\begin{array}{ll}
0 & 0 < x < a \\
\infty & \text{otherwise}
\end{array}
\right.
$$
In the given case,

$$
a = 30
$$

Hence, the energy eigen values are given by

$$
E_n = \frac{n^2 \pi^2}{ 2 \cdot 900}
$$
