# Measurement in Quantum Computation

So far, we have:
- Built quantum circuits
- Observed quantum states using Statevector

But in real experiments, we do NOT see amplitudes.

👉 We only observe measurement outcomes.

In this notebook, we will:
- Understand measurement
- Use classical registers
- Connect probabilities to outcomes

## Quantum vs Classical Information

- Quantum state → hidden (not directly observable)
- Measurement → gives classical result (0 or 1)

This is why we need:
👉 ClassicalRegister

```python
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister
from qiskit.quantum_info import Statevector
```

## Creating a Measurable Circuit

```python
QuantumCircuit?
```

```python
q = QuantumRegister(1, 'q')
c = ClassicalRegister(1, 'c')

qc = QuantumCircuit(q, c)

qc.h(q[0])   # create superposition
qc.measure(q[0], c[0])

qc
```

```python
qc.draw('text')
```

## What Do We Expect?

After H gate:

State = (|0⟩ + |1⟩)/√2

👉 Measurement outcomes:
- 0 with probability 0.5
- 1 with probability 0.5

But a single measurement gives ONLY one result.

## Simulating Measurement Using Statevector

We can extract probabilities from the statevector.

```python
# Create statevector before measurement
qc_no_measure = QuantumCircuit(1)
qc_no_measure.h(0)

sv = Statevector.from_label('0')
sv_final = sv.evolve(qc_no_measure)

# Get probabilities
probs = sv_final.probabilities()

probs
```

## Interpretation

The output shows probabilities:

[0.5, 0.5]

👉 This predicts measurement statistics.

## Repeated Measurement (Conceptual)

In real experiments:
- We repeat the circuit many times
- Collect outcomes

This gives a distribution of results.

```python
import numpy as np

# Simulate 100 measurements
samples = np.random.choice([0, 1], size=100, p=probs)

# Count results
unique, counts = np.unique(samples, return_counts=True)

dict(zip(unique, counts))
```

## Observation

- Results are approximately balanced
- Not exactly equal (randomness!)

👉 This mimics real quantum experiments.

## Key Concepts

- Statevector → gives probabilities
- Measurement → gives one outcome
- ClassicalRegister → stores outcome

Quantum → Classical transition happens during measurement.
