# Atomic spectra

There are two kinds of atomic spectra

- absorption spectra
- emission spectrum

Early experiments on absorption spectra indicate that there are specific lines of absorption. These indicate that light is absorbed in specific energy by atoms.

These experiments were enhanced by applying external electric and magnetic fields leading to the Zeeman effect and Stark effect. Under the influence of the fields, the lines are further split into sub-lines.

These experiments needed theroetical basis which was provided by Bohr. Bohr proposed the quantum atom model to account for the lines of absorption of H atom.
