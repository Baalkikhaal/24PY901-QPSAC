# Hilbert Space

Quantum states live in a mathematical structure called a Hilbert space.

We begin by recalling vectors in ℝ³ and generalizing to complex vector spaces.

## Inner Product

For vectors |ψ⟩ and |φ⟩:

⟨φ|ψ⟩ ∈ ℂ

Properties:
1. Conjugate symmetry
2. Linearity
3. Positive definiteness

|ψ⟩ → ket  
⟨ψ| → bra  
⟨φ|ψ⟩ → inner product

```python
import numpy as np

psi = np.array([1/np.sqrt(2), 1/np.sqrt(2)])
phi = np.array([1, 0])

inner = np.vdot(phi, psi)
inner
```
