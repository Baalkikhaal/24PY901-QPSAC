# Time evolution postulate

iħ d|ψ⟩/dt = H|ψ⟩

# Evolution operator

|ψ(t)⟩ = U(t)|ψ(0)⟩  
U(t) = e^{-iHt/ħ}

```python
# Two level system
import numpy as np
from scipy.linalg import expm

H = np.array([[1,0],[0,-1]])
t = 1
U = expm(-1j * H * t)
U
```
