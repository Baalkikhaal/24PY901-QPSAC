## What is a Qubit?

A classical bit can be either:
- 0
- 1

A quantum bit (qubit) can be:
- |0⟩
- |1⟩
- A combination (superposition) of both

In this notebook, we will not yet compute the state explicitly — we will focus on building circuits.

```python
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister
```

## Registers: Organizing Qubits and Bits

In Qiskit:

- QuantumRegister → stores qubits
- ClassicalRegister → stores measurement results

Think of them as containers.

```python
# Create a quantum register with 1 qubit
q = QuantumRegister(1, name='q')

# Create a classical register with 1 bit
c = ClassicalRegister(1, name='c')

# Create a quantum circuit using both registers
qc = QuantumCircuit(q, c)

qc
```

## Observation

- The circuit is currently empty.
- It contains qubits and classical bits, but no operations.

👉 Important:
A quantum circuit stores **operations**, not the quantum state itself.

## Applying Quantum Gates

We now apply operations (gates) to qubits.

Some basic gates:
- X gate → flips |0⟩ to |1⟩
- H gate → creates superposition

```python
# Apply X gate
qc.x(q[0])

qc
```

## Measurement

To extract information, we measure the qubit.

Measurement maps:
quantum → classical

```python
qc.measure(q[0], c[0])

qc
```

## Key Concept

- QuantumRegister → holds qubits
- ClassicalRegister → holds measurement outcomes
- QuantumCircuit → stores operations (gates + measurement)

The circuit is like a **program**, not the result.
