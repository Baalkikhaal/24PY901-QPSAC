# Dirac notation of Hermitian operator

Write Dirac Notation for the Hermitian operator

$$
			\tilde{Q}
			=
			\left(
			\begin{matrix}
				\alpha 	&	\gamma	\\
				\gamma^* 	&	\beta
			\end{matrix}
			\right)
            ,
$$

where $\alpha, \beta, \gamma \in \mathcal{C}$.

# Solution

The matrix can be decomposed into linear combination of projection operators.

The projection operators are

$$
\left(
\begin{matrix}
    1 	&	0	\\
    0 	&	0
\end{matrix}
\right)
,
\quad
\left(
\begin{matrix}
    0 	&	1	\\
    0 	&	0
\end{matrix}
\right)
,
\quad
\left(
\begin{matrix}
    0 	&	0	\\
    1 	&	0
\end{matrix}
\right)
,
\quad
\left(
\begin{matrix}
    0 	&	0	\\
    0 	&	1
\end{matrix}
\right)
,
$$

Thus, the operator $\tilde{Q}$ is given by

$$
\tilde{Q}
=
\alpha
\left(
\begin{matrix}
    1 	&	0	\\
    0 	&	0
\end{matrix}
\right)
+
\gamma
\left(
\begin{matrix}
    0 	&	1	\\
    0 	&	0
\end{matrix}
\right)
+
\gamma^*
\left(
\begin{matrix}
    0 	&	0	\\
    1 	&	0
\end{matrix}
\right)
+
\beta
\left(
\begin{matrix}
    0 	&	0	\\
    0 	&	1
\end{matrix}
\right)
$$

The projection operators can be written in the Dirac notation as

$$
\left(
\begin{matrix}
    1 	&	0	\\
    0 	&	0
\end{matrix}
\right)
= \ket{0}\bra{0}
$$

since

$$
\ket{0} = 
\left(
\begin{matrix}
    1 		\\
    0 	
\end{matrix}
\right)
,
\qquad
\bra{0} = 
\left(
\begin{matrix}
    0 & 1		\\
\end{matrix}
\right)
$$

Similarly

$$
\left(
\begin{matrix}
    0 	&	1	\\
    0 	&	0
\end{matrix}
\right)
= \ket{0}\bra{1}
,
\quad
\left(
\begin{matrix}
    0 	&	0	\\
    1 	&	0
\end{matrix}
\right)
=\ket{1}\bra{0}
,
\left(
\begin{matrix}
    0 	&	0	\\
    0 	&	1
\end{matrix}
\right)
= \ket{1}\bra{1}.
$$

Therefore, the operator $\tilde{Q}$ is given by

$$
\tilde{Q}
=
\alpha
\ket{0}\bra{0}
+
\gamma
\ket{0}\bra{1}
+
\gamma^*
\ket{1}\bra{0}
+
\beta
\ket{1}\bra{1}
$$
